```python
# ==========================================
# 1. Import Library
# ==========================================
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import os
import gdown
import zipfile

# ==========================================
# 2. Download dan Ekstrak Dataset
# ==========================================
file_id = '1qSI7ZqWSTBdbI1I3cwLNIOiFV-kPcNQ1'
gdown.download(f'https://drive.google.com/uc?id={file_id}', 'dataset.zip', quiet=False)

# Ekstrak file zip
with zipfile.ZipFile('dataset.zip', 'r') as zip_ref:
    zip_ref.extractall('/content/dataset')

# ==========================================
# 3. Deteksi Otomatis Folder Dataset
# ==========================================
# Tampilkan isi folder setelah ekstraksi
print("Isi folder /content/dataset:")
print(os.listdir('/content/dataset'))

# Temukan folder yang berisi subfolder train/val/test
dataset_root = '/content/dataset'
found_path = None

for root, dirs, files in os.walk(dataset_root):
    if all(sub in dirs for sub in ['train', 'val', 'test']):
        found_path = root
        break

if found_path is None:
    raise FileNotFoundError("Folder dengan struktur train/val/test tidak ditemukan. Periksa hasil ekstraksi.")

print(f"\nDataset ditemukan di: {found_path}")

# ==========================================
# 4. Menentukan Direktori Data
# ==========================================
train_dir = os.path.join(found_path, 'train')
val_dir = os.path.join(found_path, 'val')
test_dir = os.path.join(found_path, 'test')

# ==========================================
# 5. Preprocessing Data
# ==========================================
train_datagen = ImageDataGenerator(rescale=1./255)
val_datagen = ImageDataGenerator(rescale=1./255)
test_datagen = ImageDataGenerator(rescale=1./255)

train_generator = train_datagen.flow_from_directory(
    train_dir,
    target_size=(224, 224),
    batch_size=32,
    class_mode='categorical',
    shuffle=True
)

val_generator = val_datagen.flow_from_directory(
    val_dir,
    target_size=(224, 224),
    batch_size=32,
    class_mode='categorical',
    shuffle=False
)

test_generator = test_datagen.flow_from_directory(
    test_dir,
    target_size=(224, 224),
    batch_size=32,
    class_mode='categorical',
    shuffle=False
)

# ==========================================
# 6. Cek Struktur Data
# ==========================================
print("\nKelas yang terdeteksi:", train_generator.class_indices)
print(f"Total data training   : {train_generator.samples}")
print(f"Total data validation : {val_generator.samples}")
print(f"Total data testing    : {test_generator.samples}")
```

```python
# ==========================================
# 7. Membuat Model MobileNet
# ==========================================
from tensorflow.keras.applications import MobileNet
from tensorflow.keras.layers import GlobalAveragePooling2D, Dense, Dropout
from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import Adam
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from sklearn.metrics import confusion_matrix, classification_report

# Load model dasar MobileNet
base_model = MobileNet(input_shape=(224, 224, 3), weights='imagenet', include_top=False)

# Bekukan layer agar tidak ikut dilatih
for layer in base_model.layers:
    layer.trainable = False

# Tambahkan lapisan Fully Connected di atasnya
x = base_model.output
x = GlobalAveragePooling2D()(x)
x = Dense(256, activation='relu')(x)
x = Dropout(0.3)(x)
x = Dense(128, activation='relu')(x)
x = Dropout(0.3)(x)
output = Dense(len(train_generator.class_indices), activation='softmax')(x)

# Buat model akhir
model = Model(inputs=base_model.input, outputs=output)

# Kompilasi model
model.compile(optimizer=Adam(learning_rate=0.0001),
              loss='categorical_crossentropy',
              metrics=['accuracy'])

model.summary()

# ==========================================
# 8. Pelatihan Model
# ==========================================
history = model.fit(
    train_generator,
    validation_data=val_generator,
    epochs=30,   # bisa disesuaikan
    verbose=1
)

# ==========================================
# 9. Evaluasi Model di Data Uji
# ==========================================
test_loss, test_acc = model.evaluate(test_generator, steps=test_generator.samples // test_generator.batch_size)
print(f"\n✅ Akurasi Pengujian (Test Accuracy): {test_acc * 100:.2f}%")

# ==========================================
# 10. Plot Akurasi dan Loss
# ==========================================
plt.figure(figsize=(12,5))

# Plot akurasi
plt.subplot(1,2,1)
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Model Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

# Plot loss
plt.subplot(1,2,2)
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.show()

# ==========================================
# 11. Confusion Matrix dan Classification Report
# ==========================================
Y_pred = model.predict(test_generator, steps=test_generator.samples // test_generator.batch_size + 1)
y_pred = np.argmax(Y_pred, axis=1)
y_true = test_generator.classes

cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(8,6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=test_generator.class_indices.keys(),
            yticklabels=test_generator.class_indices.keys())
plt.title('Confusion Matrix')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.show()

print("\nClassification Report:")
print(classification_report(y_true, y_pred, target_names=test_generator.class_indices.keys()))
```
